```python
import pandas as pd
titanic=pd.read_csv("data/Titanic-Dataset.csv") #importing titanic data
titanic
```

```python
titanic.info()
```

```python
ages=titanic["Age"] #how to select a specific column in a dataset
ages.head()
```

```python
age_sex=titanic[["Age","Sex"]] #to select to specific columns in a dataset.
age_sex.head()
```

```python
type(titanic[["Age", "Sex"]]) #to check the data type of the returned data .We will review.
```

```python
age_sex.shape #to know number of rows and columns
```

```python
above_35=titanic[titanic["Age"]>35] #checking passengers above 35 years
above_35.head
```

```python
class_23=titanic[titanic["Pclass"].isin([2,3])] #check titanic passengers from cabin class 2 and 3
class_23.head()
```

```python
age_no_na=titanic[titanic["Age"].notna()] #passenger which age is known
age_no_na.head()
```

```python
age_no_na.shape
```

```python
adult_names=titanic.loc[titanic["Age"]>35,"Name"] #names of passengers older than 35.
adult_names.head()
```

```python
class_23=titanic[(titanic["Pclass"]==2)|(titanic["Pclass"]==3)] #using the
class_23.head()
```

```python
age_is_na=titanic[titanic["Age"].isna()] #to return a list of passengers with ages missing. NaN in pandas means null. isna() is the method used to return rows with the missing values in pandas.
age_is_na.head()
```

```python
age_is_na.shape
```

```python
adult_names=titanic.loc[titanic["Age"]>=18,["Name", "Age"]]#.loc is used for selecting specific rows and columns
adult_names.head()
```
